# Ollama Embedding

```python
from langchain_ollama import OllamaEmbeddings

# 1.创建Embedding模型
ollama_embeddings = OllamaEmbeddings(
    model="qwen3-embedding:0.6b",  # 性价比高的模型
    dimensions=1024  # 向量维度，可选：减少维度以节省存储
)

# 向量化单条文本
text = "我爱上班"
vector = ollama_embeddings.embed_query(text)
```

```python

# 批量向量化
texts = ["我要躺平", "我爱工作", "拒绝加班"]
vectors = ollama_embeddings.embed_documents(texts)
print(f"\n批量向量化: {len(vectors)} 条, 维度: {len(vectors[0])}")
```

```python
print('==============衡量相似度===============')
import numpy as np

def cosine_similarity(vec1, vec2):
    """余弦相似度函数"""
    dot = np.dot(vec1, vec2)
    return dot / (np.linalg.norm(vec1) * np.linalg.norm(vec2))

# 比较向量相似度
for i, v in enumerate(vectors):
    similarity = cosine_similarity(vector, v)
    print(f"{texts[i]}:", similarity)
```

# DashScope Embedding

```python
from langchain_community.embeddings import DashScopeEmbeddings
from dotenv import load_dotenv

load_dotenv()

# 初始化DashScopeEmbeddings，只需要指定模型名，会自动取.env读取DASHSCOPE_API_KEY
dash_scope_embeddings = DashScopeEmbeddings(
    model="text-embedding-v4",
)


# 向量化单条文本
text = "我爱上班"
vector = dash_scope_embeddings.embed_query(text)

# 批量向量化
texts = ["我要躺平", "我爱工作", "拒绝加班"]
vectors = dash_scope_embeddings.embed_documents(texts)


print('==============衡量相似度===============')
# 比较向量相似度
for i, v in enumerate(vectors):
    similarity = cosine_similarity(vector, v)
    print(f"{texts[i]}:", similarity)
```
