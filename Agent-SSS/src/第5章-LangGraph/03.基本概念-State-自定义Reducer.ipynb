```python
from operator import add
from typing import TypedDict, Annotated
from langgraph.constants import START, END
from langgraph.graph import StateGraph

# 定义Reducer函数
def add_reducer(old, new):
    return old + new

# ============1.State=================
class CustomReducerState(TypedDict):
    count: int
    nodes: Annotated[list[str], add]

# ============2.Node=================
def node_1(state: CustomReducerState):
    print(f"node_1 接收到了 State：{state}")
    return {
        "count": state['count'] + 1,
        "nodes": ['node_1']
    }

def node_2(state: CustomReducerState):
    print(f"node_2 接收到了 State：{state}")
    return {
        "count": state['count'] + 1,
        "nodes": ['node_2']
    }

# ============3.Edge连接Node，形成Graph=================
builder = StateGraph(CustomReducerState)

builder.add_node("n1", node_1)
builder.add_node("n2", node_2)

builder.add_edge(START, "n1")
builder.add_edge("n1", "n2")
builder.add_edge("n2", END)

graph = builder.compile()

# ============4.调用Graph=================
result = graph.invoke({"count": 0, "nodes": []})

print(result)
```
