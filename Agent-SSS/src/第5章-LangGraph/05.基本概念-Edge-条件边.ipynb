```python
from typing import TypedDict, Literal
from IPython.display import Image, display
from langgraph.constants import START, END
from langgraph.graph import StateGraph

# ============1.State=================
class ScoreState(TypedDict):
    score: int
    result: str

# ============2.Node=================
def score_node(state: ScoreState):
    score = input("请输入用户分数:")
    return {"score": int(score)}

def success_node(state: ScoreState):
    return {"result": "恭喜，考试通过！"}

def fail_node(state: ScoreState):
    return {"result": "节哀，考试没通过！"}


# 3.定义一个条件边判断函数，决定下个节点的名称，需要在返回值类型中，通过Literal声明可能的下一节点名
def router(state: ScoreState) -> Literal["success", "fail"]:
    score = state.get('score', 0)
    if score >= 60:
        return "success"
    else:
        return "fail"

# ============3.Edge连接Node，形成Graph=================
builder = StateGraph(ScoreState)

builder.add_node("score", score_node)
builder.add_node("success", success_node)
builder.add_node("fail", fail_node)

builder.add_edge(START, "score")
builder.add_conditional_edges("score", router)
builder.add_edge("success", END)
builder.add_edge("fail", END)

graph = builder.compile()

display(Image(graph.get_graph(xray=True).draw_mermaid_png()))
```

```python
# ============4.调用Graph=================
result = graph.invoke({})

print(result)
```
